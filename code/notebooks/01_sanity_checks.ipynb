Classical hitting times, Szegedy unitarity, and the discriminant-matrix identity.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import numpy as np

from src.classical import hitting_times, mean_hitting_time, stationary_distribution
from src.graphs import complete_graph, cycle_graph, transition_matrix
from src.szegedy import (
    check_unitarity,
    discriminant_matrix,
    spectral_gap,
    walk_matrix,
)
```

## Classical hitting times

```python
for n in (8, 16, 32):
    P = transition_matrix(complete_graph(n))
    print(n, hitting_times(P, [0])[1], mean_hitting_time(P, [0]))
```

```python
for n in (8, 16, 32):
    P = transition_matrix(cycle_graph(n))
    print(n, mean_hitting_time(P, [0]), n * (n + 1) / 6)
```

## Szegedy walk

```python
P = transition_matrix(cycle_graph(5), self_loop=0.1)
W = walk_matrix(P)
print('unitary:', check_unitarity(W))
print('discriminant block:', np.allclose(W[:5, :5], discriminant_matrix(P)))
print('spectral gap:', spectral_gap(P))
```
