# 02 Classical vs Quantum Scaling (E1)

Search time versus graph size on complete graphs, cycles, grids, and hypercubes.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

import numpy as np

from src.classical import mean_hitting_time
from src.graphs import complete_graph, cycle_graph, transition_matrix
from src.metrics import power_law_fit, success_metrics
from src.szegedy import success_probability_curve
```

```python
rows = []
for n in (8, 16, 32):
    P = transition_matrix(cycle_graph(n))
    classical = mean_hitting_time(P, [0])
    curve = success_probability_curve(P, [0], max_steps=400)
    quantum = success_metrics(curve)['t_opt']
    rows.append((n, classical, quantum))

for row in rows:
    print(row)
```

```python
sizes = np.array([row[0] for row in rows], dtype=float)
classical_times = np.array([row[1] for row in rows], dtype=float)
quantum_times = np.array([row[2] for row in rows], dtype=float)
print('classical exponent:', power_law_fit(sizes, classical_times)['exponent'])
print('quantum exponent:', power_law_fit(sizes, quantum_times)['exponent'])
```
